```python
file_name= "../dm1_25_26_dataset/DM1_game_dataset.csv"
df = pd.read_csv(file_name)
```

```python
community_features = [
    "NumWeightVotes",
    "NumAlternates",
    "NumImplementations",
    "NumExpansions"
]
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Definizione delle feature
community_features = [
    "NumWeightVotes",
    "NumAlternates",
    "NumImplementations",
    "NumExpansions"
]

# 2. Configurazione dello stile estetico
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(2, 4, figsize=(22, 11)) # Leggermente più largo per far spazio al testo
fig.subplots_adjust(hspace=0.5, wspace=0.3)

# 3. Loop per generare i plot
for i, col in enumerate(community_features):
    # Calcolo skewness dati originali
    skew_orig = df[col].skew()
    
    # --- RIGA 1: DATI ORIGINALI ---
    sns.histplot(df[col], kde=True, ax=axes[0, i], color='#3498db', bins=30)
    axes[0, i].set_title(f"Original: {col}\n(Skewness: {skew_orig:.2f})", 
                         fontsize=12, fontweight='bold')
    axes[0, i].set_xlabel("")
    axes[0, i].set_ylabel("Frequency")
    
    # Calcolo dati trasformati e relativa skewness
    log_data = np.log1p(df[col])
    skew_log = log_data.skew()
    
    # --- RIGA 2: DATI TRASFORMATI (log1p) ---
    sns.histplot(log_data, kde=True, ax=axes[1, i], color='#e67e22', bins=30)
    axes[1, i].set_title(f"Log-transformed: {col}\n(Skewness: {skew_log:.2f})", 
                         fontsize=12, fontweight='bold')
    axes[1, i].set_xlabel("Value (log scale)")
    axes[1, i].set_ylabel("Frequency")

# 4. Titolo principale
plt.suptitle("Quantitative Impact of Log Transformation on Skewness", 
             fontsize=20, fontweight='bold', y=.98)

plt.show()
```
